````python
# =============================================================================
# CELL 0: UNIVERSAL ENVIRONMENT DETECTION & SETUP
# Automatically detects: Colab / Kaggle / JupyterLab / VPS / Local
# =============================================================================

import os
import sys
import platform
from pathlib import Path

def detect_and_setup_environment():
    """Universal environment detector and configurator"""
    
    env_config = {
        'platform': 'unknown',
        'is_cloud': False,
        'has_gpu': False,
        'base_path': str(Path.cwd()),
        'supports_drive': False,
        'project_paths': [],
        'setup_strategy': 'local',
        'gpu_info': {},
        'memory_info': {}
    }
    
    print("🔍 UNIVERSAL ENVIRONMENT DETECTION")
    print("=" * 50)
    
    # =================================================================
    # 1. GOOGLE COLAB DETECTION
    # =================================================================
    try:
        import google.colab
        env_config.update({
            'platform': 'colab',
            'is_cloud': True,
            'base_path': '/content',
            'supports_drive': True,
            'setup_strategy': 'clone',
            'project_paths': [
                '/content/drive/MyDrive/ACOS',
                '/content/drive/MyDrive/ACOS-ASLI', 
                '/content'
            ]
        })
        
        print("🔗 GOOGLE COLAB detected")
        
        # Mount Google Drive
        try:
            from google.colab import drive
            drive.mount('/content/drive')
            print("✅ Google Drive mounted successfully")
            
            # Create project directories
            backup_dirs = [
                '/content/drive/MyDrive/ACOS',
                '/content/drive/MyDrive/ACOS_V52_BACKUP',
                '/content/drive/MyDrive/ACOS_V52_RESULTS'
            ]
            for backup_dir in backup_dirs:
                os.makedirs(backup_dir, exist_ok=True)
            print("📁 Backup directories ready")
            
        except Exception as e:
            print(f"⚠️  Drive mount failed: {e}")
        
        return env_config
        
    except ImportError:
        pass
    
    # =================================================================
    # 2. KAGGLE DETECTION
    # =================================================================
    if os.path.exists('/kaggle'):
        env_config.update({
            'platform': 'kaggle',
            'is_cloud': True,
            'base_path': '/kaggle/working',
            'supports_drive': False,
            'setup_strategy': 'clone',
            'project_paths': ['/kaggle/working', '/kaggle/input']
        })
        
        print("🏆 KAGGLE NOTEBOOKS detected")
        return env_config
    
    # =================================================================
    # 3. JUPYTER/JUPYTERLAB DETECTION
    # =================================================================
    try:
        from IPython import get_ipython
        if get_ipython() is not None:
            env_config.update({
                'platform': 'jupyter',
                'is_cloud': False,
                'setup_strategy': 'local_or_clone'
            })
            print("📓 JUPYTER/JUPYTERLAB detected")
    except ImportError:
        pass
    
    # =================================================================
    # 4. VPS/CLOUD SERVER DETECTION
    # =================================================================
    hostname = platform.node().lower()
    
    if any(cloud in hostname for cloud in ['aws', 'ec2', 'compute']):
        env_config.update({
            'platform': 'aws',
            'is_cloud': True,
            'setup_strategy': 'clone',
            'project_paths': ['/home/ubuntu/ACOS-ASLI', '/home/ec2-user/ACOS-ASLI']
        })
        print("☁️  AWS EC2 detected")
        return env_config
        
    elif any(cloud in hostname for cloud in ['gcp', 'google', 'gce']):
        env_config.update({
            'platform': 'gcp',
            'is_cloud': True,
            'setup_strategy': 'clone',
            'project_paths': ['/home/user/ACOS-ASLI', '/opt/ACOS-ASLI']
        })
        print("☁️  GOOGLE CLOUD detected")
        return env_config
        
    elif any(cloud in hostname for cloud in ['azure', 'microsoft']):
        env_config.update({
            'platform': 'azure',
            'is_cloud': True,
            'setup_strategy': 'clone',
            'project_paths': ['/home/azureuser/ACOS-ASLI']
        })
        print("☁️  MICROSOFT AZURE detected")
        return env_config
        
    elif any(vps in hostname for vps in ['droplet', 'linode', 'vultr']):
        env_config.update({
            'platform': 'vps',
            'is_cloud': True,
            'setup_strategy': 'clone',
            'project_paths': ['/root/ACOS-ASLI', '/home/user/ACOS-ASLI']
        })
        print("☁️  VPS SERVER detected")
        return env_config
    
    # =================================================================
    # 5. LOCAL DEVELOPMENT DETECTION  
    # =================================================================
    env_config.update({
        'platform': 'local',
        'is_cloud': False,
        'setup_strategy': 'local_path',
        'project_paths': [
            'd:/laragon/www/ACOS-ASLI',
            'D:/laragon/www/ACOS-ASLI',
            '/home/user/ACOS-ASLI',
            str(Path.home() / 'ACOS-ASLI'),
            str(Path.cwd().parent),
            str(Path.cwd())
        ]
    })
    
    print("💻 LOCAL DEVELOPMENT detected")
    return env_config

def detect_hardware_info(env_config):
    """Detect GPU, memory, and system information"""
    
    print("\n🔍 Hardware Detection...")
    
    # GPU Detection
    try:
        import torch
        if torch.cuda.is_available():
            gpu_count = torch.cuda.device_count()
            gpu_name = torch.cuda.get_device_name(0)
            gpu_props = torch.cuda.get_device_properties(0)
            gpu_memory = gpu_props.total_memory / 1024**3
            
            # Determine GPU tier
            if 'T4' in gpu_name:
                gpu_tier = 'SMALL'
            elif 'L4' in gpu_name:
                gpu_tier = 'MEDIUM'
            elif any(x in gpu_name for x in ['V100', 'A100', 'MI300X']):
                gpu_tier = 'LARGE'
            else:
                gpu_tier = 'MEDIUM' if gpu_memory >= 20 else 'SMALL'
            
            env_config.update({
                'has_gpu': True,
                'gpu_info': {
                    'name': gpu_name,
                    'memory_gb': round(gpu_memory, 2),
                    'count': gpu_count,
                    'tier': gpu_tier,
                    'sm_count': gpu_props.multi_processor_count
                }
            })
            
            print(f"🎮 GPU: {gpu_name} ({gpu_memory:.1f}GB) - Tier {gpu_tier}")
        else:
            print("💻 CPU-only execution")
    except Exception:
        print("❓ GPU detection unavailable")
    
    # Memory Detection
    try:
        import psutil
        memory = psutil.virtual_memory()
        ram_gb = memory.total / 1024**3
        available_gb = memory.available / 1024**3
        
        env_config['memory_info'] = {
            'total_gb': round(ram_gb, 1),
            'available_gb': round(available_gb, 1),
            'usage_percent': memory.percent
        }
        
        print(f"💾 RAM: {ram_gb:.1f}GB total, {available_gb:.1f}GB available")
    except Exception:
        print("❓ Memory detection unavailable")
    
    return env_config

def install_dependencies(env_config):
    """Platform-aware dependency installation"""
    
    print("\n📦 Installing dependencies...")
    
    dependencies = [
        'pytorch-crf', 'transformers', 'huggingface_hub',
        'seaborn', 'scikit-learn', 'matplotlib', 'pandas', 
        'boto3', 'tqdm', 'openpyxl', 'tabulate'
    ]
    
    if env_config['platform'] in ['colab', 'kaggle']:
        # Cloud platforms with !pip
        import subprocess
        result = subprocess.run([
            sys.executable, '-m', 'pip', 'install', '-q'
        ] + dependencies, capture_output=True, text=True)
        
        if result.returncode == 0:
            print("✅ Dependencies installed successfully")
        else:
            print("⚠️  Some dependencies may have failed")
            
    elif env_config['platform'] == 'jupyter':
        # Jupyter notebook environment
        try:
            from IPython import get_ipython
            get_ipython().system(f'pip install -q {" ".join(dependencies)}')
            print("✅ Dependencies installed via Jupyter")
        except Exception:
            print(f"💡 Please run: pip install {' '.join(dependencies)}")
            
    else:
        # Local/VPS - provide instruction
        print("💡 To install dependencies, run:")
        print(f"pip install {' '.join(dependencies)}")

# =============================================================================
# MAIN EXECUTION
# =============================================================================

print("🚀 ACOS V5.2 UNIVERSAL SETUP")
print("=" * 60)

# Detect environment
env_config = detect_and_setup_environment()

# Detect hardware
env_config = detect_hardware_info(env_config)

# Install dependencies
install_dependencies(env_config)

# Print comprehensive summary
print("\n" + "=" * 60)
print("📋 ENVIRONMENT SUMMARY")
print("=" * 60)
print(f"Platform        : {env_config['platform'].upper()}")
print(f"Cloud Platform  : {'Yes' if env_config['is_cloud'] else 'No'}")
print(f"Base Path       : {env_config['base_path']}")
print(f"Setup Strategy  : {env_config['setup_strategy']}")

if env_config.get('has_gpu') and env_config.get('gpu_info'):
    gpu = env_config['gpu_info']
    print(f"GPU             : {gpu['name']} ({gpu['memory_gb']}GB)")
    print(f"GPU Tier        : {gpu['tier']}")

if env_config.get('memory_info'):
    mem = env_config['memory_info']
    print(f"RAM             : {mem['total_gb']}GB total")

print(f"Project Paths   : {len(env_config['project_paths'])} candidates")
print(f"Drive Support   : {'Yes' if env_config['supports_drive'] else 'No'}")

# Store config for subsequent cells
globals()['ENV_CONFIG'] = env_config

print("\n🎯 Universal setup complete!")
print("💡 ENV_CONFIG available for adaptive configuration")
print("🔧 bert_utils import issue will be auto-fixed in Cell 3")
````
