## Step 3 — Library yang Tersedia

Jalankan cell di bawah untuk melihat daftar library utama yang sudah terinstall di virtual environment Anda. Library ini dapat langsung digunakan tanpa instalasi tambahan.

> **Catatan:** Jika ingin menginstall library tambahan, gunakan `pip install nama_library` di terminal JupyterLab.

```python
import sys, importlib
print(f'Python : {sys.version.split()[0]}')
print(f'Venv   : {sys.prefix}\n')

libraries = [
    ('torch',        'PyTorch — Deep Learning'),
    ('torchvision',  'TorchVision — Computer Vision'),
    ('numpy',        'NumPy — Komputasi Numerik'),
    ('pandas',       'Pandas — Analisis Data'),
    ('matplotlib',   'Matplotlib — Visualisasi'),
    ('seaborn',      'Seaborn — Visualisasi Statistik'),
    ('sklearn',      'Scikit-learn — Machine Learning'),
    ('scipy',        'SciPy — Komputasi Ilmiah'),
    ('cv2',          'OpenCV — Computer Vision'),
    ('PIL',          'Pillow — Pengolahan Citra'),
    ('transformers', 'HuggingFace Transformers — NLP/LLM'),
    ('xgboost',      'XGBoost — Gradient Boosting'),
    ('lightgbm',     'LightGBM — Gradient Boosting'),
    ('plotly',       'Plotly — Visualisasi Interaktif'),
    ('ipywidgets',   'IPyWidgets — Widget Interaktif'),
]

print(f'{{Library:<20}} {{Versi:<15}} Deskripsi')
print('-' * 65)
for mod, desc in libraries:
    try:
        m = importlib.import_module(mod)
        ver = getattr(m, '__version__', 'tersedia')
        print(f'✅ {mod:<18} {ver:<15} {desc}')
    except ImportError:
        print(f'❌ {mod:<18} {"tidak ada":<15} {desc}')

# Cek CUDA
try:
    import torch
    cuda = f'CUDA {torch.version.cuda}' if torch.cuda.is_available() else 'Tidak tersedia'
    print(f'\nGPU    : {torch.cuda.get_device_name(0) if torch.cuda.is_available() else "Tidak ada"}')
    print(f'CUDA   : {cuda}')
except: pass
```
